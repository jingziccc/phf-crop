## PHF-Crop interactive fusion webpage

Runs one current PHF-Crop trace and writes a self-contained HTML page with the multiscale fusion view. This is the only webpage-export notebook.

```python
from pathlib import Path
import os
import sys

def find_repo() -> Path:
    for candidate in (Path.cwd().resolve(), *Path.cwd().resolve().parents):
        if (candidate / "code" / "top1_hotspot").is_dir():
            return candidate
    raise RuntimeError("Run this notebook from the PHF-Crop checkout or one of its subdirectories.")

REPO = find_repo()
os.chdir(REPO)
VIZ_DIR = REPO / "visualization" / "mainline"
if str(VIZ_DIR) not in sys.path:
    sys.path.insert(0, str(VIZ_DIR))

from phf_crop_loop_panel_utils import (
    LoopPanelConfig,
    build_realtime_trace,
    display_html_multiscale_fusion_view,
    display_stage_boxes_preview,
    display_trace_text_summary,
)
```

```python
MODEL = "llava"
DATASET = "vstar"
GPU = "0"
SAMPLE_INDEX = 120
TOPMASS = 0.55
LOCALIZATION_STRATEGY = "phf"  # "phf" or "iterative_local"
SAVE = True  # writes the standalone HTML file
PREVIEW_MAX_SIDE = 1024  # inline notebook preview only; never saved
PREVIEW_LEGEND_WIDTH = 1600  # sharper separate legend; never saved
SAVE_DIR = REPO / "outputs" / "visualization" / "loop_panels"

CONFIG = LoopPanelConfig(
    repo=REPO,
    model=MODEL,
    task=DATASET,
    sample_index=SAMPLE_INDEX,
    gpu=GPU,
    topmass=TOPMASS,
    localization_strategy=LOCALIZATION_STRATEGY,
    generate_answers=True,
    save=SAVE,
    save_dir=SAVE_DIR,
)
```

```python
if "build_realtime_trace" not in globals() or "CONFIG" not in globals():
    from pathlib import Path
    import os
    import sys

    def find_repo() -> Path:
        for candidate in (Path.cwd().resolve(), *Path.cwd().resolve().parents):
            if (candidate / "code" / "top1_hotspot").is_dir():
                return candidate
        raise RuntimeError("Run this notebook from the PHF-Crop checkout or one of its subdirectories.")

    REPO = find_repo()
    os.chdir(REPO)
    VIZ_DIR = REPO / "visualization" / "mainline"
    if str(VIZ_DIR) not in sys.path:
        sys.path.insert(0, str(VIZ_DIR))

    from phf_crop_loop_panel_utils import (
        LoopPanelConfig,
        build_realtime_trace,
        display_html_multiscale_fusion_view,
        display_stage_boxes_preview,
        display_trace_text_summary,
    )
    MODEL = globals().get("MODEL", "llava")
    DATASET = globals().get("DATASET", "vstar")
    GPU = globals().get("GPU", "0")
    SAMPLE_INDEX = globals().get("SAMPLE_INDEX", 120)
    TOPMASS = globals().get("TOPMASS", 0.55)
    LOCALIZATION_STRATEGY = globals().get("LOCALIZATION_STRATEGY", "phf")
    SAVE = globals().get("SAVE", True)
    PREVIEW_MAX_SIDE = globals().get("PREVIEW_MAX_SIDE", 1024)
    PREVIEW_LEGEND_WIDTH = globals().get("PREVIEW_LEGEND_WIDTH", 1600)
    SAVE_DIR = globals().get("SAVE_DIR", REPO / "outputs" / "visualization" / "loop_panels")
    CONFIG = LoopPanelConfig(
        repo=REPO,
        model=MODEL,
        task=DATASET,
        sample_index=SAMPLE_INDEX,
        gpu=GPU,
        topmass=TOPMASS,
        localization_strategy=LOCALIZATION_STRATEGY,
        generate_answers=True,
        save=SAVE,
        save_dir=SAVE_DIR,
    )

trace, context, adapter = build_realtime_trace(CONFIG)
print("sample:", trace["row"].get("id"))
print("adaptive stop:", trace.get("adaptive_stop"))

# Dedicated text summary: question, NoCrop result, and PHF-Crop mainline result.
display_trace_text_summary(trace)

# Inline low-resolution preview only; this is not saved.
display_stage_boxes_preview(
    trace,
    CONFIG,
    max_side=PREVIEW_MAX_SIDE,
    legend_width=PREVIEW_LEGEND_WIDTH,
)

# Generates one self-contained HTML page and displays its link.
display_html_multiscale_fusion_view(trace, CONFIG)
```
